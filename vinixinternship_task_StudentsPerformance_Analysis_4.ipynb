```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries Imported Successfully!")
```

```python
url = "https://raw.githubusercontent.com/ApnaClassroom/Dataset/main/Students%20Performance.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
df.head(15)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df = df.dropna()
print("Data Cleaning Completed!")
print("Final Dataset Shape:", df.shape)
```

```python
df[["math score", "reading score", "writing score"]].mean()
```

```python
df[["math score", "reading score", "writing score"]].mean().plot(kind="bar")
plt.title("Average Student Performance")
plt.ylabel("Average Marks")
plt.xlabel("Subjects")
plt.show()
```

```python
df.to_csv("Cleaned_Students_Data.csv", index=False)
print("Cleaned Dataset Saved Successfully!")
```
